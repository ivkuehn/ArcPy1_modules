# Forest Stock Demo (interactive)

This notebook is an interactive conversion of `ForestStockDemo.py` for ENV 859. Use the dropdown to choose a tree type and the slider to choose age (0–20 years). The output shows the annual carbon stock change (m tons C / acre per year) between the current age class and the next age class.

```python
# stocking rates: stocking rate at 0,5,10,15,20 years in m tons C/acre
stockRateDict = {}
stockRateDict['loblolly']     = (14.4, 20.9, 27.2, 32.1, 37.1)
stockRateDict['oak']         = (13.4, 18.7, 24.1, 28.6, 33.2)
stockRateDict['cottonwood']  = (16.8, 21.7, 26.5, 31.1, 35.8)
stockRateDict['cypress']     = (16.7, 20.0, 26.2, 31.8, 37.2)

def get_age_class(age):
    """Return the age class index for given age in years (0..20)."""
    if age <= 5:
        return 0
    elif age <= 10:
        return 1
    elif age <= 15:
        return 2
    elif age <= 20:
        return 3
    else:
        raise ValueError('Age must be between 0 and 20 (inclusive).')

def calc_flux(tree_type, age):
    """Calculate annual flux (difference between current and next age-class stock)."""
    if tree_type not in stockRateDict:
        raise KeyError(f"Unknown tree type: {tree_type}")
    age_class = get_age_class(age)
    init_stock = stockRateDict[tree_type][age_class]
    end_stock  = stockRateDict[tree_type][age_class + 1]
    rate = end_stock - init_stock
    return init_stock, end_stock, rate
```

```python
# Interactive widgets using ipywidgets. Run this cell to interact.
import ipywidgets as widgets
from IPython.display import display, Markdown

tree_options = ['loblolly', 'oak', 'cottonwood', 'cypress']
tree_widget = widgets.Dropdown(options=tree_options, value='loblolly', description='Tree:')
age_widget = widgets.IntSlider(value=8, min=0, max=20, step=1, description='Age (yr):')

output = widgets.Output()

def on_change(change=None):
    with output:
        output.clear_output()
        tree = tree_widget.value
        age = age_widget.value
        try:
            init_stock, end_stock, rate = calc_flux(tree, age)
            display(Markdown(f"**Result**: Your **{age}** year old **{tree}** tree is storing **{rate:0.2f} m tons C/acre per year** (from {init_stock:.1f} → {end_stock:.1f})."))
        except Exception as e:
            display(Markdown(f"**Error:** {str(e)}"))

# wire up handlers and show
tree_widget.observe(on_change, names='value')
age_widget.observe(on_change, names='value')

display(widgets.HBox([tree_widget, age_widget]))
display(output)

# call once to initialize
on_change()
```
